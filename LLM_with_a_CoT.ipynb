```python
problems = [ { "question": "Sarah has 12 apples. She buys 8 more and gives 5 to her friend. How many apples does she have now?", "answer": "15" },
            { "question": "A train travels at 60 km/h for 3 hours. How far does it travel?", "answer": "180" },
            { "question": "A store sells pencils for $2 each. John buys 7 pencils. How much does he pay?", "answer": "14" },
            { "question": "There are 24 students in a class. They are divided equally into groups of 6. How many groups are formed?", "answer": "4" },
            { "question": "Emma had 50 candies. She ate 12 and then bought 20 more. How many candies does she have now?", "answer": "58" } ]
```

```python
def ask_llm(question, cot=False):
    if cot:
        prompt = f"""
Solve the following math word problem.
Think step by step and then provide the final answer.

Problem:
{question}
"""
    else:
        prompt = f"""
Solve the following math word problem and provide only the final numeric answer.

Problem:
{question}
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text
```

```python
def extract_number(text):
    numbers = re.findall(r"-?\d+\.?\d*", text)
    return numbers[-1] if numbers else None
```

```python

cot_correct = 0
no_cot_correct = 0

print("=" * 60)
print("LLM Math Reasoning Comparison")
print("=" * 60)


for i, item in enumerate(problems, start=1):

    question = item["question"]
    true_answer = item["answer"]


    no_cot_response = ask_llm(question, cot=False)


    cot_response = ask_llm(question, cot=True)


    no_cot_answer = extract_number(no_cot_response)
    cot_answer = extract_number(cot_response)


    no_cot_ok = no_cot_answer == true_answer
    cot_ok = cot_answer == true_answer

    if no_cot_ok:
        no_cot_correct += 1

    if cot_ok:
        cot_correct += 1


    print(f"\nProblem {i}")
    print(f"Question: {question}")
    print(f"Correct Answer: {true_answer}")
    print(f"No CoT Output:\n{no_cot_response}")
    print(f"CoT Output:\n{cot_response}")
    print(f"No CoT Correct: {no_cot_ok}")
    print(f"CoT Correct: {cot_ok}")


print("\n" + "=" * 60)
print(
    f"No CoT Accuracy: {no_cot_correct}/{len(problems)} "
    f"({100 * no_cot_correct / len(problems):.1f}%)"
)

print(
    f"CoT Accuracy: {cot_correct}/{len(problems)} "
    f"({100 * cot_correct / len(problems):.1f}%)"
)

print("=" * 60)
```
